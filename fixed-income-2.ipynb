```python
df.columns = (
    df.columns
        .str.lower()
        .str.strip()
        .str.replace(":", "", regex = False)
        .str.replace(" ", "_", regex = False)
        .str.replace("/", "_and_", regex = False)
        .str.replace("-", "_", regex = False)
        .str.replace("(", "", regex = False)
        .str.replace(")", "", regex = False)
        .str.replace("%", "pct", regex = False)
)
```

```python
df.columns
```

```python
df.dtypes
```

```python
df["maturity"] = pd.to_datetime(df["maturity"])
```

```python
df["issue_date"] = pd.to_datetime(df["issue_date"])
```
